```python
from pyspark.sql import functions as F
from pyspark.sql import types as T
```

```python
cus_df = spark.table("rainbow.customer_data")
cus_df.display()
```

```python
# Method - 1

def make_it_lower(name: str):
    return name.lower()

make_it_lower_udf = F.udf(make_it_lower, T.StringType())
```

```python
df = cus_df.withColumn("city", make_it_lower_udf(F.col("city")))
df.display()
```

```python
# Method 2
@udf(T.StringType())
def make_it_lower2(name: str):
    return name.lower()
```

```python
df = cus_df.withColumn("city", make_it_lower2(F.col("city")))
df.display()
```

```python
from decimal import Decimal

@udf(T.DecimalType(10,2))
def add_gst(spend_amount: int):
    return Decimal(spend_amount) * Decimal(0.18)
```

```python
df = cus_df.withColumn("gst", add_gst(F.col("amount")))
df.display()
```
