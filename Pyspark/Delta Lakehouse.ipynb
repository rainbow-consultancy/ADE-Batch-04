```python
from pyspark.sql import functions as F
from delta.tables import DeltaTable
```

```python
df = spark.read.csv("dbfs:/mnt/QuantCloudContainer/csv/apple_sales", header=True, inferSchema=True)
df.display()
```

```python
df.write.format("delta").mode("overwrite").save("dbfs:/mnt/QuantCloudContainer/Delta/apple_sales")
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("rainbow.apple_sales")
```

```python
df = (
    df
    .withColumn("dummy", F.lit("dummy_value"))
)
df.display()
```

```python
df.write.format("delta").mode("append").option("mergeSchema", "true").save("dbfs:/mnt/QuantCloudContainer/Delta/apple_sales")
```

```python
new_df = spark.read.format("delta").load("dbfs:/mnt/QuantCloudContainer/Delta/apple_sales")
new_df.display()
```

```python
dummy_null = new_df.filter(F.col("dummy").isNull()).count()
dummy_record = new_df.filter(F.col("dummy").isNotNull()).count()

print(f"dummy_null - {dummy_null}")
print(f"dummy_record_count - {dummy_record}") 
```
