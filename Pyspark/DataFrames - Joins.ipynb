```python
print("Spark DataFrames Part - 5 - Joins")
```

```python
cus_df = spark.table("samples.tpch.customer")
cus_df.display()
```

```python
ord_df = spark.table("samples.tpch.orders")
ord_df.display()
```

```python
# join the orders df and customer df to get the order information

joined_df = (
    cus_df
    .join(
        ord_df,
        on=cus_df.c_custkey == ord_df.o_custkey,
        how="left"
    )
)

joined_df.display()
```

```python
# join the orders df and customer df to get the order information

joined_df = (
    cus_df.select("c_custkey", "c_name", "c_phone")
    .join(
        ord_df.select("o_orderkey", "o_custkey"),
        on=cus_df.c_custkey == ord_df.o_custkey,
        how="left"
    )
    .drop("o_custkey")
)

joined_df.display()
```

```python
from pyspark.sql import functions as F
```

```python
# part - 2

result_df = (
    joined_df
    .groupBy("c_custkey", "c_name", "c_phone")
    .agg(
        F.count(F.col("o_orderkey")).alias("total_orders")
    )
)

result_df.display()
```

```python
# derive a column called is_vip_cus, if a customer has placed total_orders > 20 then True else False
```

```python
result_df = result_df.drop("is_vip_cus")
```

```python
result_df = (
    result_df
    .withColumn(
        "is_vip_cus", 
        F.when(F.col("total_orders") > 20, "SuperValued")
        .when(F.col("total_orders") > 10, "Valued")
        .otherwise("Regular")
    )
)
result_df.display()
```
