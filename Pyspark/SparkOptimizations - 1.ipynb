```python
spark.conf.set("spark.sql.adaptive.enabled", "false")
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1)
```

```python
from pyspark.sql import functions as F
```

```python
# Spark Join Stratagies

# 1. Sort Merge Join  - default
# 2. Shuffel Hash join -
# 3. Broadcast Join - 
```

```python
cus_df = spark.table("samples.tpch.customer")
cus_df.display()
```

```python
ord_df = spark.table("samples.tpch.orders")
ord_df.display()
```

```python
cus_df.rdd.getNumPartitions()
```

```python
ord_df.rdd.getNumPartitions()
```

```python
# join the orders df and customer df to get the order information

joined_df = (
    cus_df
    .join(
        ord_df,
        on=cus_df.c_custkey == ord_df.o_custkey,
        how="left"
    )
)

joined_df.display()
```

```python
joined_df.count()
```

```python
ord_df.count()
```

```python
cus_df.count()
```

```python
# join the orders df and customer df to get the order information

broadcast_df = (
    ord_df
    .join(
        F.broadcast(cus_df),
        on=cus_df.c_custkey == ord_df.o_custkey,
        how="left"
    )
)

broadcast_df.display()
```

```python
spark.conf.set("spark.sql.adaptive.enabled", "true")
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", 314572800)
```

```python
# auto broadcast the orders df and customer df to get the order information

auto_df = (
    cus_df
    .join(
        ord_df,
        on=cus_df.c_custkey == ord_df.o_custkey,
        how="left"
    )
)

auto_df.display()
```
