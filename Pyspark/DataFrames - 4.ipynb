```python
from pyspark.sql import functions as F
```

```python
apple_df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("/FileStore/tables/apple_global_sales_dataset.csv")
)
display(apple_df)
```

```python
df = (
    apple_df
    .groupBy("country")
    .agg(
        F.sum(F.col("units_sold")).alias("total_units_sold"),
        (F.sum(F.col("revenue_usd"))).alias("total_revenue_usd"),
        F.round(F.sum(F.col("revenue_usd"))).alias("total_revenue_usd"),
        F.round(F.sum(F.col("revenue_usd")), 2).alias("total_revenue_usd"),
        F.ceil(F.sum(F.col("revenue_usd"))).alias("total_revenue_usd"),
        F.floor(F.sum(F.col("revenue_usd"))).alias("total_revenue_usd")
    )
)

df.display()
```

```python
# In a given year I want to find out how many products where Kept and how many were Exchanged and how many were Returned

df = (
    apple_df
    .groupBy("year", "return_status")
    .agg(
        F.count("*").alias("count_all")
    )
    .orderBy("year")
)

df.display()
```

```python
df = (
    apple_df
    .groupBy(
        F.year("sale_date"), F.month("sale_date")
    )
    .agg(
        F.round(F.sum(F.col("revenue_usd")), 2).alias("total_revenue")
    )
    .orderBy(F.year("sale_date"), F.month("sale_date"))
)

df.display()
```
