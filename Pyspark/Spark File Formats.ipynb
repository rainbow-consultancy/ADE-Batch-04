```python
df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("dbfs:/mnt/QuantCloudContainer/csv/apple_global_sales_dataset.csv")
)

df.display()
```

```python
df.rdd.getNumPartitions()
```

```python
df.write.format("csv").mode("overwrite").save("dbfs:/mnt/QuantCloudContainer/csv/apple_sales")
```

```python
df_repartioned = df.repartition(5)
df_repartioned.write.format("csv").mode("overwrite").save("dbfs:/mnt/QuantCloudContainer/csv/apple_sales")
```

```python
df_repartioned = df.coalesce(1)
df_repartioned.write.format("csv").option("header", True).mode("overwrite").save("dbfs:/mnt/QuantCloudContainer/csv/apple_sales")
```

```python
df = spark.read.csv("dbfs:/mnt/QuantCloudContainer/csv/apple_sales", header=True, inferSchema=True)
df.display()
```
