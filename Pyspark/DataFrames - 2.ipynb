```python
# Read a table from Delta Sahres

trips_df = spark.table("samples.nyctaxi.trips")
trips_df.display()
```

```python
# - withColumn function
from pyspark.sql import functions as F

trips_df = trips_df.withColumn("fare_amount_inr", F.col("fare_amount") * 92)
trips_df.display()
```

```python
trips_df = (
    trips_df
    .withColumn("fare_amount_inr", F.col("fare_amount") * 92)
    .withColumn("dummy1", F.lit("dummy"))
    .withColumn("dumm2",F.lit("dummy"))
)
trips_df.display()
```

```python
# withColumnRenamed

trips_df = trips_df.withColumnRenamed("fare_amount_inr", "fare_amount_rupee")
trips_df.display()
```

```python
trips_df = spark.sql("select * from samples.nyctaxi.trips")
trips_df.display()
```

```python
# method-1

df = trips_df.select("tpep_pickup_datetime", "tpep_dropoff_datetime")
df.show()
```

```python
# method - 2
df = trips_df.select(trips_df.tpep_pickup_datetime, trips_df.tpep_dropoff_datetime)
df.show()
```

```python
# method - 3
df = trips_df.select(trips_df["tpep_pickup_datetime"], trips_df["tpep_dropoff_datetime"])
df.show()
```

```python
# method - 4  -- prefered method
df = trips_df.select(F.col("tpep_pickup_datetime"), F.col("tpep_dropoff_datetime"))
df.show()
```
